```python
import torch
from anysolv import predict_solvation_energy
from anysolv.solvent import _solvents_ci
from ase.build import molecule
from fairchem.core import FAIRChemCalculator
from fairchem.core.units.mlip_unit import load_predict_unit
from fairchem.core.units.mlip_unit.api.inference import (
    inference_settings_default,
)
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)
```

```python
inference_settings_uma = inference_settings_default()
inference_settings_uma.compile = False
```

```python
predictor = load_predict_unit(
    "../models/uma-s-1p2.pt",
    inference_settings=inference_settings_uma,
    device=device,
    seed=41,
)
calc = FAIRChemCalculator(predictor, task_name="omol")
```

```python
atoms = molecule("CH4")
atoms.info["charge"] = 0
atoms.info["spin"] = 1
atoms.calc = calc
```

```python
delta_energy_solvent, delta_forces_solvent = predict_solvation_energy(
    atoms,
    charge=atoms.info["charge"],
    spin=atoms.info["spin"],
    solvent="dmso",
    checkpoint="../models/model_moe.pt",
    device=device,
    inference_settings="default",
)
```
